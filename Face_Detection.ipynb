```python
import os
import cv2
import numpy as np
from ultralytics import YOLO
from IPython.display import display
from PIL import Image
```

```python
folder = "photos/1"
folders = {0: "photos/1", 1:"photos/2" }


face_model = YOLO("face_yolov8m.pt")
face_recognizer = cv2.face.LBPHFaceRecognizer_create()
faces = []
labels = []

for label, folder in folders.items():
    files = os.listdir(folder)
    for file in files:
        photo = cv2.imread(folder + "/" + file)
        face_result = face_model(photo, verbose=False)
        processed_image = face_result[0].plot()

        left, top, right, bottom = face_result[0].boxes.xyxy[0].int()
        face = photo[top:bottom, left:right]
        face = cv2.cvtColor(face, cv2.COLOR_BGR2GRAY)
        face = cv2.resize(face, (200,200))
        faces.append(face)
        labels.append(label)

face_recognizer.train(faces, np.array(labels))
face_recognizer.write("face_recognizer.yml")
# Image.fromarray(faces[0][:, :, ::-1])
#Image.fromarray(faces[0])
```

```python
cap = cv2.VideoCapture(2)

names = {0: "Messi", 1:"Ronaldo"}
max_distance = 70

face_model = YOLO("face_yolov8m.pt")
face_recognizer = cv2.face.LBPHFaceRecognizer_create()
face_recognizer.read("face_recognizer.yml")

while cv2.waitKey(1) != ord("x"):
    _, frame = cap.read()
    face-result = face-model(frame, verbose=False)
    processed_feed = face_result[0].plot()

    for box in face_result[0].boxes.xyxy:
        left, top, right, bottom = box.int()
        face = frame[top:bottom, left:right]
        face = cv2.cvtColor(face, cv2.COLOR_BGR2GRAY)
        face = cv2.resize(face, (200,200))
        pred_label, distance = face_recognizer.predict(face)

        if distance > max_distance:
            name = "Unknown"
        else:
            name = names[pred_label]
        cv2.putText(processed_feed, name, (int(left), int(bottom) + 20), 0, 0.8, (255, 255, 255), 2)

        cv2.imshow("my window", processed_feed)

cv2.waitKey(5000)
cap.release()
cv2.destroyAllWindows()
```
