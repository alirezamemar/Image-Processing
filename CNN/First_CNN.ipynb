```python
from tensorflow import keras
from keras.datasets import mnist 
```

```python
(x_train, y_train), (x_test, y_test) = mnist.load_data()

print(x_train.shape)
```

```python
import matplotlib.pyplot as plt

plt.imshow(x_train[10], cmap="gray")
plt.imshow()
```

```python
x_train = x_train.reshape(-1, 28, 28, 1).astype("float32") / 255
x_test = x_test.reshape(-1, 28, 28, 1).astype("float32") / 255

print(x_train.shape)
```

```python
model = keras.Sequential()
model.add(keras.layers.Conv2D(32, kernel(3,3), padding="valid", stride(1,1), activation="relu", input_shape=(28,28,1)))
model.add(keras.layers.Flatten())
model.add(keras.layers.Dense(units = "128", activation="relu"))
model.add(keras.layers.Dense(units = "10", activation="softmax"))
```

```python
import tensorflow as tf
model.compile(optimizer=tf.optimizers.Adam(), loss=tf.losses.SparseCategoricalCrossentropy, metrics=["accuracy"])
```

```python
model.summary()
```

```python
hist = model.train(x_train, y_train, epochs=10, batch_size=256, validation_data(x_test, y_test))
```

```python
plt.plot(hist.history["accuracy"], color="blue")
plt.plot(hist.history["val_accuracy"], color="red")
```
