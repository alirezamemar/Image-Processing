```python
import cv2 as cv

# Read image basic(BRG)
image = cv.imread("./output.jpg")
cv.imshow("basic", image)
cv.waitKey()
print(image.shape)
```

```python
image_gray = cv.cvtColor(image, cv.COLOR_BGR2GRAY)
cv.imshow("gray", image_gray)
cv.waitKey()
cv.destroyAllWindows()
```

```python
# Read image gray_scale
image = cv.imread("./output.jpg", 0)
cv.imshow("GRAY_SCALE", image)
cv.waitKey()
cv.destroyAllWindows()
print(image.shape)
```
